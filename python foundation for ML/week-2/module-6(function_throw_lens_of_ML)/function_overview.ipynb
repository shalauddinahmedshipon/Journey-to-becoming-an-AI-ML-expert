```python
def print_values(*args):
    for i in args:
        print(i)

print_values(10,30,20)
```

```python
def adder(**key_args):
    for key,val in key_args.items():
        print(f"{key}:{val}")

adder(a=1,b=2,c=3)
```

```python
st={10,20,30}
st_iter=iter(st)
print(next(st_iter))
print(next(st_iter))
print(next(st_iter))
```

generator

```python
lst=[x for x in range(500)]
def data_loader(chunk_size,lst):
    for i in range(0,len(lst),chunk_size):
        yield lst[i:i+chunk_size]
x=data_loader(5,lst)
print(next(x))
print(next(x))
print(next(x))
print(next(x))
```
