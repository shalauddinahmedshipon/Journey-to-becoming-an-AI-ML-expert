## Practice 1 :

Get the text from sample.txt , and count the following


Number of lines

Number of words

Number of characters

and save it to the counter_of_string.txt file

```python
from functools import reduce
with open("./sample_data/sample.txt","r") as file:
  string_list=file.readlines()
  string_list=list(map(lambda x:x.strip(),string_list))
  print(string_list)
  # Number of lines
  number_of_lines=len(string_list)
  print(number_of_lines)

   # Number of word
  words_list=list(map(lambda str:str.split(),string_list))
  print(words_list)
  words_len=list(map(lambda x:len(x),words_list))
  print(words_len)
  number_of_words=reduce(lambda x,y:x+y,words_len)
  print(number_of_words)
  #Number of characters
  sentence_list=list(map(lambda str: str.replace(" ",""),string_list))
  print(sentence_list)
  sentence_len=list(map(lambda x:len(x),sentence_list))
  print(sentence_len)
  number_of_character=reduce(lambda x,y:x+y,sentence_len)
  print(number_of_character)

  #write the result in the new file
  with open("./sample_data/counter_of_string.txt","w") as file:
    file.write(f"number of lines:{number_of_lines}\nnumber of words:{number_of_words}\nnumber of character:{number_of_character}")
```
