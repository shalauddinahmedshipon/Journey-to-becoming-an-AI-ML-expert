```python
class Phone:
  category="Electronics"    ##common attribute for all object called class attribute
  ## Constractor
  def __init__(self,model,battery,camera,battery_percentage=100):    ##different attribute for all different object called instant attribute
    self.model=model
    self.battery=battery
    self.camera=camera
    self.battery_percentage=battery_percentage

  ## method
  def charge(self,hour):
    self.charge+=hour

  def capture(self,photo):
    if(self.battery_percentage<=0):
      print("No Photo Capture")
    else:
      self.battery_percentage-=photo
      print(f"Photo capture in {self.model}")

## creating object

apple=Phone("iphone 20",3000,50)
samsung=Phone("note 50",4000,50)
nokia=Phone("nokia 1100",5000,10)



print(apple.model)
print(samsung.category)
print(nokia.camera)

print(apple.battery_percentage)
apple.capture(10)
print(apple.battery_percentage)
```
