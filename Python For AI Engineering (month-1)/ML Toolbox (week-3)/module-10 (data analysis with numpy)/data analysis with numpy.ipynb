## NDarry

```python
import numpy as np

arr1=np.array([1,2,3])
print(type(arr1))
print(arr1.ndim)
```

```python
arr2=np.array([[1,2,3],[4,5,6]])
print(arr2.ndim)
```

```python
arr3=np.array([[[1,2,3],[4,5,6],[5,1,9]],[[1,2,3],[4,5,6],[5,1,9]]])
print(arr3.ndim)
```

## Attribute of NDarry

```python
##dimention
print(arr1.ndim)
print(arr2.ndim)
print(arr3.ndim)

##shape
print(arr1.shape)
print(arr2.shape)
print(arr3.shape)

## datatype
print(arr1.dtype)
print(arr2.dtype)
print(arr3.dtype)


##size
print(arr1.size)
print(arr2.size)
print(arr3.size)

```

## NDarry datatype

```python
arr=np.array([1,2,3])
print(arr.dtype)
```

```python
#upcasted to float
arr=np.array([1,2.5,3])
print(arr.dtype)
```

```python
#upcasted to string
arr=np.array([1,2.5,3,'h'])
print(arr.dtype)
```

```python
#selecting a datatype for an array
arr=np.array([1,2,500],dtype=np.int32)
print(arr.dtype)
arr=arr.astype(np.int16)
print(arr.dtype)
print(arr)
```

## NDarray creation from existing data

```python
## list
lst=[10,20,30,40,50.5]
arr=np.array(lst,dtype=np.int32)
print(arr)
print(type(arr))
print(arr.dtype)
```

```python
mix_lst=[10,False,50.5,'hello']
arr=np.array(mix_lst)
print(arr)
print(type(arr))
print(arr.dtype)
```

```python
## matrix

matrix=[
    [1,2,3],
    [4,5,6]
]

arr=np.array(matrix)
print(arr)
print(type(arr))
print(arr.dtype)
print(arr.ndim)
print(arr.shape)
```

```python
## tuple
tp=(1,2,4)
arr=np.array(tp)
print(arr)
print(type(arr))
print(arr.dtype)
print(arr.ndim)
print(arr.shape)
```

```python
## set
st={1,2,3,4}
arr=np.array(list(st))
print(arr.dtype)
print(arr)
arr=arr.astype(np.int16)
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
```

```python
## dictonary
dc={'a':10,'b':20,'c':30}
keys = dc.keys()
values=dc.values()
items=dc.items()
# print(items)
```

```python
# Keys to ndarray
arr=np.array(list(keys))
print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)
```

```python
# values to ndarray
arr=np.array(list(values))
print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)
```

```python
#items to ndarray
arr=np.array(list(items))
print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)
```

## array creation with random values

```python
# np.random.rand(shape)

arr=np.random.rand(10)  #0 -> 1
print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)
```

```python
# np.random.randint(start,end,shape)

arr=np.random.randint(10,20,(3,5)) 
print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)
```

```python
# np.random.uniform(start,end,shape)
arr=np.random.uniform(1,20,(50,5)) 
print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)
```

## array creation with range functions

```python
#np.arange(first,end,step) -> similar to range function

arr=np.arange(1,10,1)
mat=arr.reshape(3,3)
print(arr)
print(mat)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)
```

```python
# np.linspace(start,end(included), how much pointer)
arr=np.linspace(0,4,6)
print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)
```

```python
#np.logspace( start, end(included) ,how much pointer)
arr=np.logspace(0,4,6,base=2)
print(arr)
```

## indexing and slicing

```python
#indexing

print(arr1)
print(arr2)

arr1[2] = 100

arr2[0][2]= 100

print(arr1)
print(arr2)
```

```python
#slicing
#1d = arr[start:end:step]
print(arr1)
arr1_mod=arr1[0:2].copy()
print(arr1_mod)
arr1_mod[1]=200
print(arr1_mod)
print(arr1)
```

```python
#2d slicing
# arr[ row_start : row_end : step , column_start : column_end : step ]
print(arr2)
```

```python
#getting a row
row_0=arr2[0:1 ,]
row_1=arr2[1: ,]
print(row_0)
print(row_1)
```

```python
#getting a column
col_0=arr2[::,0:1]
col_1=arr2[::,1:2 ]
print(col_0)
print(col_1)
```

```python
#getting a portion
portion=arr2[::,1:]
print(portion)
```
