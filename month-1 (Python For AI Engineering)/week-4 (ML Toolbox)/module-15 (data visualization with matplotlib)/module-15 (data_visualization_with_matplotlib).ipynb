# Scatter Plot

```python
students = pd.read_csv('student_IQdata.csv') 
students
```

```python
#Relation between Study Hours and IQ Score - postive correlation

plt.xlabel('Study_Hour') 
plt.ylabel('IQ_Score') 
plt.title('Relation between Study Hours and IQ Score') 

plt.scatter(students['Study_Hour'], students['IQ_Score'])
```

```python
#Relation between Shoe size and IQ Score - > no correlation

plt.xlabel('Shoe_Size') 
plt.ylabel('IQ_Score') 
plt.title('Relation between Shoe Size and IQ Score') 

plt.scatter(students['Shoe_Size'], students['IQ_Score'])
```

```python
#Relation between Chilling hour and IQ Score - > negative correlation

plt.xlabel('Chilling_Hours') 
plt.ylabel('IQ_Score') 
plt.title('Relation between Chilling_Hours and IQ Score') 

plt.scatter(students['Chilling_Hours'], students['IQ_Score'],marker='d')
```

# Bar Chart 

```python
data = {
    'Student': [ 'B', 'C', 'D', 'E', 'F', 'G', 'H'],
    'Gender': [ 'Male', 'Female', 'Male', 'Female', 'Male', 'Female', 'Male'],
    'StudyHours': [ 2, 5, 3, 7, 4, 8, 1],
    'Attendance': [ 60, 88, 70, 95, 80, 98, 55],
    'Grade': [ 'Fail', 'Pass', 'Fail', 'Pass', 'Pass', 'Pass', 'Fail']
} 

df = pd.DataFrame(data) 
df
```

```python
# gender wise 

gender_group = df.groupby('Gender').size() 

gender_group
```

```python
gender_group.index
```

```python
gender_group.values
```

```python
plt.xlabel('Gender') 
plt.ylabel('Count') 
plt.title('Bar Chart of Gender among students') 

plt.bar(gender_group.index, gender_group.values,color=['pink','black'])
```

```python
#  Bar chart on real data 

df = pd.read_csv('student_data.csv') 

df
```

```python
plt.xlabel('Completion Status') 
plt.ylabel('Count') 
plt.title('Bar Chart of Course completion')

status = df.groupby('CompletionStatus').size() 
plt.bar(status.index,status.values,color=['green','yellow','red'])
plt.show() 
```

```python
# location 
location = df.groupby('Location').size() 

plt.bar(location.index,location.values)

plt.xlabel('Location') 
plt.ylabel('Count') 
plt.title('Bar Chart of Location of Students')
plt.show() 
```

# pie chart

```python
status = df.groupby('CompletionStatus').size() 
plt.pie(status,labels=status.index,autopct='%1.1f%%',shadow=True,explode=(0,0,0.2))
plt.show()
```
