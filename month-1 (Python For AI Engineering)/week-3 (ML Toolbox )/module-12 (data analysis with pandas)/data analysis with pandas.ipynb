## Different File Loading in Pandas

```python
# csv file

csv_data = pd.read_csv("student_data.csv")
print(csv_data)
print(type(csv_data))
```

```python
# excel file

excel_file = pd.read_excel('phitron_student_marks.xlsx') 
print(excel_file)
print(type(excel_file))
```

```python
#parquet file
parquet_file = pd.read_parquet('students.parquet') 
print(parquet_file)
print(type(parquet_file))
```

```python
# json file
json_file = pd.read_json("data.json")
print(json_file)
print(type(json_file))
```

## Transforming to df from inbuilt datas

```python
my_list = [['Alice', 25], ['Bob', 30], ['Charlie', 28]]
list_df = pd.DataFrame(my_list,columns=['Name','Age'],index=[1,2,3])
list_df
```

```python
my_tuple = (('Alice', 25), ('Bob', 30), ('Charlie', 28))
tuple_df= pd.DataFrame(my_tuple,columns=['Name','Age'],index=[1,2,3])
tuple_df
```

```python
my_dict = {
    'Name': ['Alice', 'Bob', 'Charlie'],
    'Age': [25, 29,28]
}

dict_df = pd.DataFrame(my_dict,index=[1,2,3])
dict_df

```

```python
my_data = [
    {'Name': 'Alice', 'Age': 25, 'City': 'New York'},
    {'Name': 'Bob', 'Age': 30, 'City': 'Paris'},
    {'Name': 'Charlie', 'Age': 28}
]

data_df= pd.DataFrame(my_data)
data_df
```
