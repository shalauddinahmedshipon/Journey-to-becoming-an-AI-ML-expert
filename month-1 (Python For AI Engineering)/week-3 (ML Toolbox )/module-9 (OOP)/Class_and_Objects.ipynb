```python
# parent / base class

class Phone:
  category="Electronics"    ##common attribute for all object called class attribute
  ## Constractor
  def __init__(self,model,battery,camera,battery_percentage=100):    ##different attribute for all different object called instant attribute
    self.model=model
    self.battery=battery
    self.camera=camera
    self.battery_percentage=battery_percentage

  ## method
  def charge(self,hour):
    print(f"Charge completed by {hour} hr")

  def capture(self,photo):
    if(self.battery_percentage<=0):
      print("No Photo Capture")
    else:
      self.battery_percentage-=photo
      print(f"Photo capture in {self.model}")

## creating object

apple=Phone("iphone 20",3000,50)
samsung=Phone("note 50",4000,50)
nokia=Phone("nokia 1100",5000,10)



print(apple.model)
print(samsung.category)
print(nokia.camera)

print(apple.battery_percentage)
apple.capture(10)
print(apple.battery_percentage)
```

```python
## child / derived class

##Single inheritance
class SmartPhone(Phone):
    def __init__(self,model,battery,camera,processor):
        super().__init__(model,battery,camera)
        self.processor=processor
    def charge(self,hr):
        print("first charging is running!")
        super().charge(hr)






pro=SmartPhone("nokia1100",2000,5000,"snapdragon")
    
print(pro.model)
print(pro.processor)
pro.charge(10)
```

```python
class Cooling_Mechanism :
    def __init__(self,cooling_method):
        self.cooling_method=cooling_method

    def cooling_on(self):
        print(f"the system is to being cool by {self.cooling_method}")
```

```python
## Multiple Inheritance
class SmartPhone_CoolingMode(SmartPhone,Cooling_Mechanism):
    def __init__(self, model, battery, camera, processor,cooling_method):
        SmartPhone.__init__(self, model, battery, camera, processor)
        Cooling_Mechanism.__init__(self,cooling_method)


pro_cooling = SmartPhone_CoolingMode("x",200,5000,"pros","nitro")

print(pro_cooling.processor)
print(pro_cooling.battery)
print(pro_cooling.cooling_method)
print(pro_cooling.cooling_on())
print(pro_cooling.charge(5))
       
```
