For loop:

```python
for i in range(10):
    print(i)
```

```python
for i in range(1,11):
    for j in range(1,6,2):
        print(i,j)
```

while loop:

```python
count=1
while count<11 :
    print(count,"Hello world!")
    count+=1
```

continue and break statement:

```python
sum=0
for i in range(11):
    if i%2==0: continue
    sum+=i
    if sum>=20: break
    print(sum)
```
